Respuesta

La única restricción del enunciado es que haya partido el viernes y partido el lunes; las coincidencias están permitidas. Se cuentan las asignaciones que la incumplen y se restan del total, por **inclusión-exclusión**:

- $A$, sin partido el viernes: los 10 partidos se reparten en los 9 horarios restantes, $|A| = 9^{10} = 3.486.784.401$.
- $B$, sin partido el lunes: $|B| = 9^{10} = 3.486.784.401$.
- $A \cap B$, sin partido ni el viernes ni el lunes: todo en los 8 horarios del fin de semana, $|A \cap B| = 8^{10} = 1.073.741.824$.

$$|A \cup B| = |A| + |B| - |A \cap B| = 5.899.826.978$$

$$\text{asignaciones válidas} = 10^{10} - 2 \cdot 9^{10} + 8^{10} = 4.100.173.022$$

**Interpretación.** «Asignar obligatoriamente un partido el viernes y un partido el lunes» se interpreta como *al menos* uno cada día, porque el enunciado permite coincidencias. Si se exigiera *exactamente* uno, quedarían $10 \cdot 9 \cdot 8^{8} = 1.509.949.440$ asignaciones. El óptimo que se obtiene más abajo tiene exactamente un partido el viernes y otro el lunes, así que cumple las dos lecturas.

**Representación de una solución.** Una solución asigna un horario a cada partido de la jornada, así que se representa con una **tupla (o lista) de longitud $n$**: la posición $i$ es el partido y el valor, su horario, por ejemplo `('S20', 'D18', 'V20', ...)`. **(*)¿Cual es la función objetivo?**

Respuesta

Para una asignación $(h_1, \dots, h_n)$ de horarios a los $n$ partidos, la audiencia total de la jornada es

$$f(h_1, \dots, h_n) = \sum_{i=1}^{n} B(c_i) \cdot K(h_i) \cdot \big(1 - P(k_i)\big)$$

donde $B(c_i)$ es la audiencia base del partido $i$ según las categorías de sus equipos, $K(h_i)$ el coeficiente de su horario, $k_i$ el número de otros partidos en el mismo horario y $P(k_i)$ la reducción por coincidencias de la tabla del enunciado. Se busca la asignación que maximiza $f$ con al menos un partido el viernes y otro el lunes.

La función `calcular_audiencia` implementa esta fórmula; más abajo se comprueba que reproduce el total de la jornada de ejemplo del enunciado (5,88 millones).

**(*)¿Es un problema de maximización o minimización?**

Respuesta

Es un problema de maximización, ya que se busca asignar los partidos a los horarios de forma que maximice la audiencia.

**Diseña un algoritmo para resolver el problema por fuerza bruta**

Respuesta

Se implementa un algoritmo por fuerza bruta que prueba todas las asignaciones posibles de horarios (`itertools.product`), descarta las que no tienen partido el viernes y el lunes y se queda con la de mayor audiencia.

Con los 10 partidos de la jornada son $10^{10}$ asignaciones y no termina en un tiempo razonable, así que se ejecuta con los $n$ primeros partidos ($n = 4, \dots, 7$): así se comprueba que funciona y se mide cómo crece su coste. Cada ejecución tiene un tiempo máximo (función `ejecutar`).

```python
# Imports
import random
import time
from collections import Counter
from itertools import product
```

```python
# Excepciones personalizadas
class TimeoutException(Exception):
    pass
```

```python
# Estructura de datos
horarios = [
    "V20",
    "S12",
    "S16",
    "S18",
    "S20",
    "D12",
    "D16",
    "D18",
    "D20",
    "L20"
]

idx_viernes = horarios.index("V20")
idx_lunes = horarios.index("L20")

categorias = ["A", "B", "C"]

# Audiencia base (millones)
audiencia_base = {
    ("A", "A"): 2.0,
    ("A", "B"): 1.3,
    ("A", "C"): 1.0,
    ("B", "A"): 1.3,
    ("B", "B"): 0.9,
    ("B", "C"): 0.75,
    ("C", "A"): 1.0,
    ("C", "B"): 0.75,
    ("C", "C"): 0.47
}

# Coeficientes por horario
coef_horario = {
    "V20": 0.4,
    "S12": 0.55,
    "S16": 0.70,
    "S18": 0.80,
    "S20": 1.00,
    "D12": 0.45,
    "D16": 0.75,
    "D18": 0.85,
    "D20": 1.00,
    "L20": 0.40
}

# Penalización por coincidencias
penalizacion = {
    0: 0.00,
    1: 0.25,
    2: 0.45,
    3: 0.60,
    4: 0.70,
    5: 0.75,
    6: 0.78,
    7: 0.80,
    8: 0.80
}
```

```python
# Funciones auxiliares
def audiencia_sin_penalizacion(partido, horario):
    return (audiencia_base[partido] * coef_horario[horario])

def calcular_audiencia(partidos, asignacion):
    conteo = Counter(asignacion)

    total = 0

    for partido, horario in zip(partidos, asignacion):

        base = audiencia_sin_penalizacion(partido, horario)

        coincidencias = conteo[horario] - 1

        reduccion = penalizacion[coincidencias]

        total += base * (1 - reduccion)

    return total

def cumple_restricciones(asignacion):
    # Verifica si la asignación cumple con las restricciones:
    tiene_viernes = horarios[idx_viernes] in asignacion # Al menos un partido el viernes
    tiene_lunes = horarios[idx_lunes] in asignacion # Al menos un partido el lunes
    return (tiene_viernes and tiene_lunes)

def ejecutar(algoritmo, partidos, segundos):
    # Ejecuta un algoritmo con un tiempo máximo y mide lo que tarda.
    # Devuelve (audiencia, asignacion, segundos); si se agota el tiempo, audiencia = None
    inicio = time.perf_counter()
    try:
        audiencia, asignacion = algoritmo(partidos, inicio + segundos)
    except TimeoutException:
        audiencia, asignacion = None, None
    return audiencia, asignacion, time.perf_counter() - inicio

def miles(x):
    # Entero con punto como separador de miles
    return f"{x:,}".replace(",", ".")

def mostrar_asignacion(partidos, asignacion):
    # Imprime cada partido de la jornada con el horario que se le asigna
    for i, (partido, horario) in enumerate(zip(partidos, asignacion), 1):
        print(f"Partido {i:>2}: {partido[0]}-{partido[1]} -> {horario}")
```

```python
# Algoritmo por fuerza bruta
def maximizar_audiencia_fuerza_bruta(partidos, limite=None):
    mejor_audiencia = -1
    mejor_asignacion = None

    # Genera por fuerza bruta todas las combinaciones posibles de horarios para los partidos:
    # 10^n combinaciones (con los 10 partidos de una jornada, 10 mil millones).
    # Cada 'asignacion' es una tupla: el índice es el partido y el valor, su horario.
    combinaciones = product(horarios, repeat=len(partidos))

    for asignacion in combinaciones:
        # Cortar si se supera el tiempo límite
        if limite is not None and time.perf_counter() > limite:
            raise TimeoutException()

        # Filtrar por restricciones
        if not cumple_restricciones(asignacion):
            continue

        # Calcular la audiencia total de esta combinación específica
        audiencia_actual = calcular_audiencia(partidos, asignacion)

        # Mantener el máximo global
        if audiencia_actual > mejor_audiencia:
            mejor_audiencia = audiencia_actual
            mejor_asignacion = asignacion

    return mejor_audiencia, mejor_asignacion
```

```python
# Jornada de descripción del problema
partidos_descripcion = [
    ('B', 'A'),
    ('B', 'A'),
    ('C', 'C'),
    ('B', 'A'),
    ('C', 'C'),
    ('B', 'C'),
    ('B', 'B'),
    ('B', 'B'),
    ('B', 'C'),
    ('B', 'B')
]
```

```python
# Comprobación de calcular_audiencia con la jornada de ejemplo del enunciado (total: 5,88)
horarios_ejemplo = ("V20", "S12", "S16", "S18", "S20", "D16", "D16", "D18", "D20", "L20")
audiencia_ejemplo = calcular_audiencia(partidos_descripcion, horarios_ejemplo)
print(f"Audiencia de la jornada de ejemplo: {audiencia_ejemplo:.4f} millones (enunciado: 5,88)")
assert round(audiencia_ejemplo, 2) == 5.88
```

```python
# Con los 10 partidos son 10^10 combinaciones: inviable (ver la complejidad más abajo).
# Se ejecuta con los n primeros partidos para comprobar que funciona y medir su coste.
tiempos_fb, optimos_fb = {}, {}

print(f"{'partidos':>8} | {'combinaciones':>13} | {'mejor audiencia':>15} | {'segundos':>8}")
print("-" * 54)
for n in range(4, 8):
    audiencia, asignacion, segundos = ejecutar(maximizar_audiencia_fuerza_bruta,
                                               partidos_descripcion[:n], 1800)
    assert audiencia is not None, "la fuerza bruta no ha terminado a tiempo"
    tiempos_fb[n], optimos_fb[n] = segundos, audiencia
    print(f"{n:>8} | {miles(10 ** n):>13} | {audiencia:>15.4f} | {segundos:>8.2f}")

print("\nMejor asignación con los 7 primeros partidos:")
mostrar_asignacion(partidos_descripcion[:7], asignacion)
```

**Calcula la complejidad del algoritmo por fuerza bruta**

Respuesta

Con $n$ partidos y $m$ horarios, la fuerza bruta genera las $m^n$ asignaciones y, para cada una, comprueba las restricciones y calcula la audiencia, las dos cosas en $O(n)$:

$$T(n) = O(n \cdot m^n)$$

Es **exponencial** en el número de partidos, con memoria $O(n)$ (solo guarda la asignación en curso y la mejor). Con la jornada completa ($n = m = 10$) son $10^{10}$ asignaciones y del orden de $10^{11}$ operaciones. La tabla anterior lo confirma: cada partido más multiplica el tiempo por unas $m = 10$ veces. La celda siguiente extrapola el tiempo medido a la jornada completa.

```python
# Extrapolación a los 10 partidos: mismo tiempo por combinación y por partido
por_combinacion = tiempos_fb[7] / 10 ** 7
estimado = por_combinacion * 10 ** 10 * (10 / 7)
print(f"Tiempo medido por combinación (n = 7): {por_combinacion * 1e6:.2f} µs")
print(f"Factor entre n = 6 y n = 7: {tiempos_fb[7] / tiempos_fb[6]:.1f}")
print(f"Estimación con los 10 partidos: {miles(round(estimado))} s ≈ {estimado / 3600:.1f} horas")
```

**(*)Diseña un algoritmo que mejore la complejidad del algortimo por fuerza bruta. Argumenta porque crees que mejora el algoritmo por fuerza bruta**

Respuesta

Se propone utilizar un algoritmo de retroceso (backtracking) con poda (bound). Este algoritmo consta de los siguientes pasos:

1. Asignar partidos uno a uno.
2. Calcular audiencia parcial.
3. Estimar una cota superior de la audiencia máxima posible.
4. Si esa cota ya es peor que la mejor solución encontrada, se poda la rama.

Con esto conseguimos:
- Evitar explorar millones de combinaciones inútiles.
- Mantener la optimalidad.
- Explorar una fracción del espacio.

```python
# Algoritmo retroceso (backtracking) con poda (bound)
nodos_visitados = {}

def maximizar_audiencia_retroceso(partidos, limite=None):
    n = len(partidos)
    nodos = 0

    # Máxima contribución teórica de cada partido
    maximos = [audiencia_base[p] for p in partidos]

    # Sumas acumuladas de máximos
    sufijos = [0] * (len(partidos) + 1)

    for i in range(len(partidos) - 1, -1, -1):
        sufijos[i] = sufijos[i + 1] + maximos[i]

    mejor_audiencia = -1
    mejor_asignacion = None

    def retroceso(indice, asignacion_actual, audiencia_parcial):
        nonlocal mejor_audiencia, mejor_asignacion, nodos
        nodos += 1

        # Poda: la audiencia parcial no descuenta coincidencias y los partidos que faltan
        # suman su máximo, así que la cota nunca subestima la audiencia alcanzable
        cota = (audiencia_parcial + sufijos[indice])

        if cota <= mejor_audiencia:
            return

        # Caso base
        if indice == n:
            # Filtrar por restricciones
            if not cumple_restricciones(asignacion_actual):
                return

            audiencia_real = calcular_audiencia(partidos, asignacion_actual)

            if audiencia_real > mejor_audiencia:
                mejor_audiencia = audiencia_real
                mejor_asignacion = asignacion_actual.copy()

            return

        # Cortar si se supera el tiempo límite
        if limite is not None and time.perf_counter() > limite:
            raise TimeoutException()

        # Expandir nodo
        partido = partidos[indice]

        for horario in horarios:
            contribucion = audiencia_sin_penalizacion(partido, horario)

            nueva_audiencia = (audiencia_parcial + contribucion)

            asignacion_actual.append(horario)

            retroceso(indice + 1, asignacion_actual, nueva_audiencia)

            asignacion_actual.pop()

    retroceso(0, [], 0)
    nodos_visitados["retroceso"] = nodos

    return mejor_audiencia, mejor_asignacion
```

```python
# Con la jornada completa esta versión supera las 2 horas en este entorno: se ejecuta con
# los 8 primeros partidos para medirla y compararla después con la versión mejorada
print("Buscando la mejor combinación por retroceso (backtracking) con poda (bound)...")
audiencia_rp, asignacion_rp, tiempo_rp = ejecutar(maximizar_audiencia_retroceso,
                                                  partidos_descripcion[:8], 1800)
assert audiencia_rp is not None, "el retroceso no ha terminado a tiempo"
nodos_rp = nodos_visitados["retroceso"]

print(f"Mejor audiencia con los 8 primeros partidos: {audiencia_rp:.3f} millones")
print("Asignación óptima:")
mostrar_asignacion(partidos_descripcion[:8], asignacion_rp)
print(f"Nodos visitados: {miles(nodos_rp)}")
print(f"Ejecución terminada en {tiempo_rp:.1f} segundos ({tiempo_rp / 60:.1f} minutos)")
```

Esta primera versión es lenta: con solo los 8 primeros partidos tarda 30 segundos y visita 21.523.291 nodos, y con la jornada completa supera las 2 horas en el entorno en que se ha ejecutado este notebook (en la ejecución original del grupo, en otro equipo, tardó unos 42 minutos). Su cota es débil porque la audiencia parcial no descuenta las coincidencias. Por este motivo se realizan las siguientes mejoras:

- Precalcular la audiencia de cada partido en cada horario.
- Ordenar los partidos de mayor a menor audiencia, para encontrar pronto una buena solución y podar más.
- Explorar primero los mejores horarios.
- Reemplazar `Counter` por listas indexadas por horario y actualizar la audiencia de forma incremental, de modo que la cota ya incluye las penalizaciones por coincidencias.

```python
# Función auxiliar
def obtener_penalizacion(conteo):
    idx = max(0, conteo - 1)
    return penalizacion.get(idx, penalizacion[max(penalizacion)])

# Algoritmo retroceso (backtracking) con poda (bound) mejorado
def maximizar_audiencia_retroceso_mejorado(partidos, limite=None):
    n = len(partidos)
    m = len(horarios)
    nodos = 0

    # Precalcular audiencias
    audiencias = [[audiencia_base[p] * coef_horario[h] for h in horarios] for p in partidos]

    # Ordenar partidos
    orden = sorted(range(n), key = lambda i: max(audiencias[i]), reverse = True)

    partidos = [partidos[i] for i in orden]
    audiencias = [audiencias[i] for i in orden]

    # Sumas acumuladas de máximos
    sufijos = [0] * (n + 1)

    for i in range(n - 1, -1, -1):
        sufijos[i] = sufijos[i + 1] + max(audiencias[i])

    mejor_audiencia = -1
    mejor_asignacion = None

    asignacion = [-1] * n

    conteo = [0] * m

    partidos_horario = [[] for _ in range(m)]

    contribucion = [0.0] * n

    audiencia_actual = 0.0

    def retroceso_mejorado(indice, tiene_viernes, tiene_lunes):
        nonlocal mejor_audiencia, mejor_asignacion, audiencia_actual, nodos
        nodos += 1

        # Poda: audiencia ya penalizada de los partidos colocados + máximo de los que faltan
        cota = (audiencia_actual + sufijos[indice])

        if cota <= mejor_audiencia:
            return

        # Caso base
        if indice == n:
            if not (tiene_viernes and tiene_lunes):
                return

            if audiencia_actual > mejor_audiencia:
                mejor_audiencia = audiencia_actual
                mejor_asignacion = asignacion.copy()

            return

        # Cortar si se supera el tiempo límite
        if limite is not None and time.perf_counter() > limite:
            raise TimeoutException()

        # Probar horarios, primero los de más audiencia
        orden_horarios = sorted(range(m), key = lambda h: audiencias[indice][h], reverse = True)

        for h in orden_horarios:
            # Guardar estado
            audiencia_anterior = audiencia_actual

            afectados = partidos_horario[h].copy()

            contribuciones_antiguas = [contribucion[p] for p in afectados]

            # Insertar partido
            partidos_horario[h].append(indice)

            conteo[h] += 1

            asignacion[indice] = h

            penal = obtener_penalizacion(conteo[h])

            # Recalcular todos los partidos del horario
            for p in afectados:
                audiencia_actual -= contribucion[p]
                contribucion[p] = (audiencias[p][h] * (1 - penal))
                audiencia_actual += contribucion[p]

            contribucion[indice] = (audiencias[indice][h] * (1 - penal))

            audiencia_actual += contribucion[indice]

            retroceso_mejorado((indice + 1), (tiene_viernes or (h == idx_viernes)),
                               (tiene_lunes or (h == idx_lunes)))

            # Deshacer
            audiencia_actual = audiencia_anterior

            partidos_horario[h].pop()

            conteo[h] -= 1

            contribucion[indice] = 0

            for p, c in zip(afectados, contribuciones_antiguas):
                contribucion[p] = c

    retroceso_mejorado(0, False, False)
    nodos_visitados["retroceso mejorado"] = nodos

    if mejor_asignacion is None:
        return None, None

    # Recuperar orden original
    resultado = [None] * n

    for pos, original in enumerate(orden):
        resultado[original] = horarios[mejor_asignacion[pos]]

    return mejor_audiencia, resultado
```

```python
print("Buscando la mejor combinación por retroceso (backtracking) con poda (bound) mejorado...")
audiencia_rpm, asignacion_rpm, tiempo_rpm = ejecutar(maximizar_audiencia_retroceso_mejorado,
                                                     partidos_descripcion, 3600)
assert audiencia_rpm is not None, "el retroceso mejorado no ha terminado a tiempo"
nodos_rpm = nodos_visitados["retroceso mejorado"]

print(f"Mejor audiencia: {audiencia_rpm:.3f} millones")
print("Asignación óptima:")
mostrar_asignacion(partidos_descripcion, asignacion_rpm)
print(f"Partidos el viernes: {asignacion_rpm.count('V20')}, "
      f"el lunes: {asignacion_rpm.count('L20')}")
print(f"Nodos visitados: {miles(nodos_rpm)}")
print(f"Ejecución terminada en {tiempo_rpm:.1f} segundos")

# Comprobaciones: la audiencia recalculada desde cero coincide y hay exactamente un partido
# el viernes y otro el lunes
assert abs(calcular_audiencia(partidos_descripcion, asignacion_rpm) - audiencia_rpm) < 1e-9
assert asignacion_rpm.count("V20") == 1 and asignacion_rpm.count("L20") == 1

# Comparación con la primera versión sobre los mismos 8 partidos: mismo óptimo, menos nodos
audiencia_8, _, tiempo_8 = ejecutar(maximizar_audiencia_retroceso_mejorado,
                                    partidos_descripcion[:8], 600)
nodos_8 = nodos_visitados["retroceso mejorado"]
assert abs(audiencia_8 - audiencia_rp) < 1e-9
print(f"\nCon los 8 primeros partidos: primera versión {tiempo_rp:.1f} s "
      f"y {miles(nodos_rp)} nodos;")
print(f"versión mejorada {tiempo_8:.1f} s y {miles(nodos_8)} nodos "
      f"({nodos_rp / nodos_8:.0f} veces menos nodos, mismo óptimo)")
```

**(*)Calcula la complejidad del algoritmo**

Respuesta

**Peor caso.** Si la cota no podara nada, el algoritmo recorrería el mismo árbol que la fuerza bruta, con $m^n$ hojas. En cada nodo ordena los $m$ horarios ($O(m \log m)$) y actualiza la audiencia de los partidos que comparten horario ($O(n)$), así que en el peor caso es

$$T(n) = O\big(m^n \, (m \log m + n)\big)$$

exponencial, como la fuerza bruta. La memoria es $O(n \cdot m)$, por la tabla de audiencias precalculadas, más la profundidad de la recursión, $O(n)$.

**En la práctica** la poda descarta casi todo el árbol, y eso es lo que mejora a la fuerza bruta. Con los 8 primeros partidos, la primera versión visita 21.523.291 nodos y la mejorada 563.961 nodos. Con la jornada completa, la versión mejorada encuentra el óptimo visitando 37.389.621 nodos en 177 segundos, frente a los más de $10^{10}$ nodos del árbol completo, que la fuerza bruta tardaría unas 18 horas en recorrer, y a las más de 2 horas de la primera versión. La mejora viene de dos cosas: la cota de la versión mejorada ya incluye las penalizaciones, así que es más ajustada, y al ordenar partidos y horarios la mejor solución aparece pronto y poda desde el principio.
